## Task 1: Mos Def's "Rock'n'Roll"

In `./texts/`, you find the file `rocknroll.txt`, which contains the lyrics to Mos Def's "Rock'n'Roll" (listen [here](https://youtu.be/Py3EpE4dE7s?si=jTCwklHo7RXw8WRR)).

1. Read in the file line by line (`.readlines()`)
2. What is the total number of lines in the file?
3. Remove all the empty lines, and lines that start with `[` (as they are not part of the lyrics)
4. Count the number of lines in the song (after removing the empty and `[`) lines
5. Count the number of words in the song (`.split()` can help!)
6. How often does the word `soul` appear in the song? (Check whether you're right by looking it up with Ctrl+F in the text file)
7. Select your three favourite (swear?) words and censor them, replacing them by a string that is of the same length, but contains only the first letter, while all other letters are replaced by `*`. (For example, if you want to censor `fight`, replace all occurences of the word `fight` by `f****`); 
8. Write the outcome to a new file, `rocknroll-censored.txt`, to be saved in the same folder as the original text file.

```python
with open("./texts/rocknroll.txt", "r") as f:
    lines = f.readlines()
print(f"Number of lines in the file: {len(lines)}")
```

```python
# removing all empty lines, and lines that start with "[":
lines_final = [] # initiate empty list
# loop through lines
for line in lines:
    if line != "\n" and line[0] != "[":
        lines_final.append(line.replace("\n", ""))
print(f"Final number of lines: {len(lines_final)}")
```

```python
wordcount = 0
for line in lines_final:
    wordcount += len(line.split())
print(f"Number of words in the song: {wordcount}")
# This is not entirely correct, because orthographic signs like "-" are being counted as words here. But that's ok.
```

```python
soul_count = 0
for line in lines_final:
    soul_count += line.lower().count("soul")
print(f"Soul appears {soul_count} times")
```

```python
# censoring
censored_words = ["money", "credit", "control"]
# new list (to be written to file later)
lines_censored = []
for line in lines_final:
    for word in censored_words:
        line = line.replace(word, word[0] + "*"*(len(word)-1))
    lines_censored.append(line + "\n")   
```

```python
with open("./texts/rocknroll-censored.txt", "w") as f:
    f.writelines(lines_censored)
```

# Task 2: All of your favourite books

In `/texts/`, you find the file `favouritebooks.txt`, which contains this class' reply to the pre-lecture01 mentimeter question "What is your favourite book?"

1. Read in the file.
2. Remove all the `\n`s at the end of each line.
3. Replace all underscores by white spaces.
4. Capitalize all letters (so that it reads "THE BELL JAR" instead of "The bell jar", etc.)
5. Which book is mentioned the most amount of times? How many times is it mentioned? 
6. How many different (unique... ;) books are on the list?

> Hot tip for task 4: To make your life much easier, check out https://docs.python.org/3/library/collections.html#collections.Counter . Then try to import the `Counter` function from the `collections` module, and call the `Counter` function with the list of books as input.

```python
# YOUR CODE HERE
with open("./texts/favouritebooks.txt", "r") as f:
    books_raw = f.readlines()

books = []
for book in books_raw:
    book = book.replace("\n", "").replace("_", " ").upper()
    books.append(book)
books
```

```python
from collections import Counter
Counter(books)
# Harry Potter, mentioned 7 times
```

```python
# unique books on the list:
len(set(books))
```

# Task 3: Smalltalk with the computer

Write the function `smalltalk()` that
* asks the user how they are doing "on a scale from 1 to 5" (`input()`)
* prints out the user's input
* prints out an appropriate reply (depending on the user input)
* asks the user whether they like the weather today
* prints out an appropriate reply (depending on whether the user said "yes" or "no")
* says goodbye to the user in a funny way

> Spoiler: We will pick this task up in the next exercise, where you will be running this code from the CLI!

```python
# Note: this task has many potential solutions; below is just one example.

def smalltalk():
    # mood_dict has possible user inputs as keys,
    # and possible replies as values
    # e.g. if user says "1", the script will reply "Sorry to hear that"
    mood_dict = {
        "1" : "Sorry to hear that.", 
        "2" : "Tomorrow is another day!",
        "3" : "I'm also doing so-so.",
        "4" : "That's good to hear!",
        "5" : "That's awesome!"
    }

    # weather_dict has possible inputs as keys,
    # and possible replies as values 
    weather_dict = {
        "yes" : "Me too!",
        "no"  : "Me neither!"
    }

    ########## ASKING ABOUT THE MOOD ############

    # first question: mood; save user input to the variable "mood"
    q1 = "How are you doing today, on a scale from 1 to 5? "
    print(q1)
    mood = input(q1)
    # print the question and the answer to the CLI
    print(mood)
    # get the answer
    ans1 = mood_dict[mood]
    # display the answer to the user
    print(ans1)

    ########## ASKING ABOUT THE WEATHER ############

    # second question: weather; save user input to the variable "weather"
    q2 = "Do you like the weather today? "
    print(q2)
    weather = input(q2)
    print(weather)
    # get the answer
    ans2 = weather_dict[weather]
    # display the answer to the user
    print(ans2, "...anyways, my tea is getting cold, BYYYYYYEEEEE!")
    
```

```python
smalltalk()
```
