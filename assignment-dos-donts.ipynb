# Common mistakes and better notebook formatting

This notebook contains errors that are inspired by the student submissions for Assignment 2.

The goal is to help students understand how to submit cleaner, more readable notebooks and avoid the most common errors.

If you'd like to learn more about Markdown syntax, you can check out https://www.markdownguide.org/basic-syntax/

## Markdown cells vs. code cells

A Jupyter notebook is made of two different kinds of cells:

### 1) Markdown cells
- Used for explanations, headings, and written text.
- Example: this cell you are reading now.
- If in VSCode, double click to edit markdown

### 2) Code cells
- Used only for Python code.
- Everything in a code cell should be valid Python.

### Why this matters
A notebook should be structured so that it is easy to read and easy to run. If code cells contain Markdown text, or if code is unfinished, the notebook can fail.

### General piece of advice for Assignments and the Exam:
 > **Before submitting, always run: `Restart` and then `Run All`. If a cell fails, fix it before you hand it in.**
___

## Mistake 1: Markdown vs Python cells

This happened when task descriptions or code examples were copied directly into a code cell with the triple backticks.

Python then reads those backticks as part of the code, which leads to a syntax error.

````python
```python
def should_i_bike(rain):
    if rain:
        return False
    else:
        return True

rain_today = False
if should_i_bike(rain_today):
    print("bike peacefully")
```
````

### Why this is a problem:
The first line starts with backticks, so Python cannot parse the cell as valid Python code.

### Fix
- Keep code cells free of Markdown fences.
- Only write actual Python code inside a code cell.

### TA recommendation:

- If there is a task that requires you to describe lines of code, it is a good idea to write comments next to the code itself.
- You can copy the code from the assignment text into a Python cell to doublecheck how it runs

## Mistake 2: Assertions before the function is defined or missing

This was a frequent issue: students wrote assert statements before the function existed.

It also often came with the wrong function name, such as `is_even` instead of `is_odd`.

````python
# This fails because the function has not been defined yet
assert is_odd(1) == True
assert is_odd(7659) == True
assert is_odd(-3) == True

# Wrong function name: the task asks for is_odd, not is_even
def is_even(n):
    return n % 2 == 0
````

### Why this is a problem
Python executes the notebook from top to bottom. When it reaches the assert lines, the function `is_odd` does not exist yet.

### Fix
- Define the function first.
- Then write the tests/assertions.
- Make sure the function name matches the assignment exactly.
- For this task, the function should be named `is_odd`.

## Mistake 3: Placeholder code left in the notebook

Some students left unfinished or placeholder code in their final submission.

This included patterns such as `# YOUR CODE HERE` or `raise NotImplementedError()`.

These are not finished solutions and should not remain in the final notebook.

````python
# Task 3: Divisibles by 7
divby7 = []
for n in range(1, 101):
    # TODO: finish this part
    raise NotImplementedError("I forgot to implement Task 3")

print(divby7)
````

### Why this is wrong
The notebook is still incomplete. The task asks for a working solution, not a placeholder.

### Fix
- Remove `raise NotImplementedError()` before submitting.

## Mistake 4: Docstring in the wrong place

- If you are using docstrings, make sure to write them right after the function definition. They will not work otherwise 

````python
def foo():
    """This is a docstring."""
    x = 1

foo?
````

````python
def foo():
    x = 1
    """This is just a string literal."""
foo?
````

## Final checklist before submission

- Every code cell contains valid Python.
- Markdown cells are used for text, explanations, and headings.
- Code cells contain only Python code.
- Functions are defined before they are used.
- There are no placeholders like `# YOUR CODE HERE` or `raise NotImplementedError()`.
- **The notebook runs from top to bottom without errors**.
- You use `Restart and Run All` before handing it in.
