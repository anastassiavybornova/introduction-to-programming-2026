```python
# import libraries
import json
```

```python
# filename as user input (will be used for riseified notebook)
filename_pre = "./lecture05-prerise.ipynb"
filename_post = filename_pre.replace("-prerise", "")

# define dict with metadata we want
riseify = {
    "deletable" : True,
    "editable": True,
    "slideshow": {
        "slide_type": "slide"
    },
    "tags": []
}

# read in notebook in text format
with open(filename_pre, "r") as file:
    content = json.load(file)

# edit notebook metadata cell by cell
for cell in content["cells"]:
    for k, v in riseify.items():
        if not k in cell["metadata"]:
            cell["metadata"][k] = v
```

```python
with open(filename_post, "w") as file:
    json.dump(content, file)
```
