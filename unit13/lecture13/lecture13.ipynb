## Subsetting DataFrames

* Boolean indexing (filtering by condition)
* bitwise operators! `& | ~` instead of ~~and or not~~

# Boolean indexing (filtering by condition). This is a powerful tool!

##### `df[columnlabel]` can be combined with comparison `> < == !=`  operators to create a `condition`
##### `conditions` can be combined with logical operators: ~~and or not~~ "bitwise and"  `&`, "bitwise or" `|`, and "not" `~`  operators
##### `df[condition]` returns only those rows where condition is True
##### `df[(condition1) & (condition2)]` returns only those rows where both conditions are True
##### `df[(condition1) | (condition2)]` returns only those rows at least one of the conditions is True
##### `df[(~condition1) & (~condition2)]` returns only those rows where both conditions are False

````python
# let's see what happens if we use a comparison operator with a single column:
# returns for each row either True or False
df["Age"] > 18 # is this person over 18?
````

````python
# we can use that condition to index only rows where condition is True:
my_condition = df["Age"] > 18 # who was over 18 on that ship?
df[my_condition]
````

````python
# a shorter (but perhaps more confusing at first) way to write this: 
# df[condition] (where condition contains a df column)
df[ df["Age"]>18 ]
````

````python
# filtering by several conditions:
# put each condition inside () round brackets
# combine them with & (meaning "and") or | (meaining "or")
# everyone that was over 18 AND survived 
df[ (df["Age"]>18) & (df["Survived"]==1) ]
````

````python
# everyone that is male and under 25
df[ (df["Sex"]=="male") & (df["Age"]<25) ]
````

````python
# everyone that is either female, or over 18, or both 
df[ (df["Sex"]=="female") | (df["Age"]>18) ]
````

## Tilde operator `~` in pandas means `not` ("inversion" operator)

````python
# everyone with an Age value below 18 (NA are not considered)
df[df["Age"] < 18]
````

````python
# everyone with an Age value above or equal 18 (NA are NOT considered)
df[df["Age"] >= 18]
````

````python
# everyone with an Age value NOT below 18 (NA are CONSIDERED!)
df[~(df["Age"] < 18)]
````

## DATA CLEANUP: DEALING WITH MISSING VALUES

### Filtering out missing values (NaNs) with boolean conditions

* `NaN` .... "not a number"
* pandas methods: `.isna()` and `.notna()` (isna = is not available, notna = not not available)
* Note, for pandas: "not available" (includes both `NaN` and `None`)

````python
df.head()
# in rows with index 0,2,4 the Cabin data is missing
````

````python
df["Cabin"].isna()
# returns for each row whether data is NOT AVAILABLE (True) or available (False)
````

````python
df["Cabin"].notna()
# returns for each row whether data is AVAILABLE (True) or not available (False)
````

### How to use `isna` and `notna` to clean data?

````python
# boolean indexing to only have rows where we know the Cabin value:
my_condition = df["Cabin"].notna()
df[my_condition]
# or, shorter: df[df["Cabin"].notna()]
````

````python
# filter by available data in *both* Age and Cabin columns 
df[(df['Age'].notna()) & (df['Cabin'].notna())]
````

````python
# calling notna on entire dataframe:
df.notna()
````

## Machine Learning 101 in 15 minutes: Monte Carlo simulation to approximate $\pi$ with Python 😱

* Monte Carlo: ML (machine learning) technique: "throw random sampling at it"
* $\pi$ is an irrational number (cannot be expressed as ratio $a/b$), and also a transcendental number (cannot be a solution of an algebraic equation involving only finite sums, products, powers, and integers) (cf. [wiki](https://en.wikipedia.org/wiki/Pi)). This means we cannot "compute" $\pi$
* But already with our Python knowledge so far, we can use the machine lerning technique "Monte Carlo" (aka random sampling) to APPROXIMATE it!


## (lecture worksheet + blackboard)

````python
# You don't need to understand this code - 
# run this cell just for visualization purposes!
import random
import matplotlib.pyplot as plt
import numpy as np

random.seed(161)

size = 100
xs = np.random.uniform(0, 1, size = size)
ys = np.random.uniform(0, 1, size = size)

fig, axs = plt.subplots(2,2, figsize=(8,8))

axs_indeces = [(0,0), (0,1), (1,0), (1,1)]

for i in axs_indeces:
    axs[i].add_patch(plt.Circle([0,0], 1, color="lightblue"))
    axs[i].set_xlim([0,1])
    axs[i].set_ylim([0,1])

for i, nr_points in zip(axs_indeces, [5,20,50,100]):
    axs[i].scatter(x = xs[:nr_points], y = ys[:nr_points], s = 2)
    axs[i].set_title(f"{nr_points} random points")
````
