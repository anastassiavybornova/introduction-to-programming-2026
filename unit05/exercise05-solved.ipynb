## Task 3: Counting letters

**(3a)** Write a for-loop that iterates over the string `"THE CAPITAL OF MADAGASCAR IS ANTANANARIVO"` and counts how many times the capital letter `A` appears in it. After the for-loop, print out the count. **Hint:** define the count before the for loop, and set it to 0.

**(3b)** Use your solution (the for-loop) from (3a) to define the function `count_A()`: 
* `count_A()` takes a string as an input; 
* uses a `for` loop to count how many times the capital letter `A` appears in the input string;
* and returns the count value. 

Test your function `count_A()` on the string `"THE CAPITAL OF MADAGASCAR IS ANTANANARIVO"` by `assert`ing that it returns the expected number of letters.

**(3c)** Use your solution from (3b) to define the function `count_Aa()`, which counts *both* capitalized and small letters (`A` and `a`). Test your modified function on the string `"A change is as good as a rest"` by asserting that it returns the expected number of letters.


````python
# Sample solution for task 3a
input_string = "THE CAPITAL OF MADAGASCAR IS ANTANANARIVO"
count = 0
for letter in input_string:
    if letter=="A":
        count +=1
print(count)
````

````python
# Sample solution for task 3b
def count_A(input_string):
    count = 0
    for letter in input_string:
        if letter=="A":
            count +=1
    return count

assert count_A("THE CAPITAL OF MADAGASCAR IS ANTANANARIVO")==10
````

````python
# Sample solution for task 3c
def count_Aa(input_string):
    count = 0
    for letter in input_string:
        if letter in ["A", "a"]:
            count +=1
    return count 

assert count_Aa("A change is as good as a rest")==5
````

## Task 5: Summing up numbers

**(5a)** Use a `for` loop and the variable `my_sum` (initally set to 0) to calculate the sum of all numbers from 0 to 500.

**(5b)** Use a `for` loop and the variable `my_sum` (initally set to 0) to calculate the sum of all **even** numbers from 0 to 500.

**(5c)** Use a `for` loop and the variable `my_sum` (initally set to 0) to calculate the sum of all **divisible by 100** numbers from 0 to 500.

> **Hint:** For all three `for`-loops, use `range()` to create the iterable object 

> **Hint:** Use the modulo operator `%` for (5b) and (5c)

````python
# 5a: sum of numbers from 0 to 500
my_sum = 0
for i in range(501): # !! to include the number 500, range() must go up to 501
    my_sum += i
print(my_sum)
````

````python
# 5b: sum of all even numbers from 0 to 500
# same as above, but using a conditional and the modulo %:
my_sum = 0
for i in range(501): # !! to include the number 500, range() must go up to 501
    if i % 2 == 0:
        my_sum += i
print(my_sum)
````

````python
# 5b: alternative solution: we can also make range() take steps of 2:
my_sum = 0
for i in range(0, 501, 2):
    my_sum += i
print(my_sum)
````

````python
# 5c: same as above, but now with modulo 100 instead of modulo 2
my_sum = 0
for i in range(501):
    if i % 100 == 0:
        my_sum += i
print(my_sum)
````

## Task 6: Finding divisors

Use the `%` operator and a `for` loop to find out by which numbers from 2 to 30 is `462231` divisible? Save them into a list and print out the list.

````python
# Numbers from 2 to 30 by which 462231 is divisible
divisors = [] # initiate an empty list
for i in range(2,31): # for each number from 2 to 30,
    if 462231 % i == 0: # if 462331 divided by this number has modulo 0,
        divisors.append(i) # append this number to the list
print(divisors) # print out the list
````

## Task 7: Summing up numbers, but we stop at 253

Use a `for` loop **or** a `while` loop to find out how many integers (0 + 1 + 2 + 3 ... + n) you need to sum up to obtain 253. (There are many different ways to do this! If you find just 1, that's enough :) 

**Hints** Think about how and when to break out of the loop; think about how to make sure that at each loop iteration, we increase by 1 the number that we are adding to the sum.

> We can (but we don't have to...) write this task as an equation, that we want to solve for `n`:

$\sum_{i=0}^{n}{i}=253$


````python
my_sum = 0
count = 0
while True:
    my_sum += count
    if my_sum >= 253:
        break # !! break *before* increasing the count to get the right integer at which to stop!
    count += 1
print(count, my_sum)
````

In the code below, somebody tried to remove all 2's from a list by looping through the list *indeces*, and removing the list *element* if its value is equal to 2. 
* Why does this not work? What is the hidden danger in this code?
* Correct the code so that it removes all `2`s from the list! (there are several ways to do this)

```python

for i in range(len(my_list)):
     if my_list[i] == 2:
        my_list.remove(my_list[i])
print(my_list)
```

````python
# Find the bug in the code below!
my_list = [2, 1, 2, 3, 4, 5]
for i in range(len(my_list)):
    if my_list[i] == 2:
        my_list.remove(my_list[i])
print(my_list)
````

````python
# The bug comes from CHANGING the list while we ITERATE over it.
# DON'T DO THIS!
# A better solution, instead of REMOVING UNWANTED numbers, is to
# create a NEW LIST and ADDING WANTED numbers:
my_list = [2, 1, 2, 3, 4, 5]
my_new_list = []
for i in range(len(my_list)):
    if my_list[i] != 2: # note that here we changed to "!=" instead of "=="
        my_new_list.append(my_list[i])
print(my_new_list)
````
