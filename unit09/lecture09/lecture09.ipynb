# Syntax for reading in & writing files

```python
### READING
with open("filepath", "r") as opened_file:
    my_text = opened_file.read() # .read() will give you 1 single string
    ## OR:
    my_text = opened_file.readlines() # .readlines() will give you 1 string for each line

### WRITING
with open("filepath", "w") as opened_file:
    opened_file.write(my_text) # if my_text is a string
    ## OR:
    opened_file.writelines(my_text) # if my_text is a list of strings
```

## Three equivalent ways of running a python script in a subfolder

The trick is to **make sure that the CLI knows where to look for the file you want to run.**

(1) Providing the absolute filepath:
```
python <absolute-filepath-to-file>
```

(2) Providing the relative filepath:
```
python <relative-filepath-to-file>
````

(3) Navigating to the exact location of the file:
```
cd <path-to-folder>
python <name-of-file>
```
