## Task 2: Slicing strings

```python
"Actions speak louder than words."
```
For the string defined above, use slicing to save each word (without white spaces) in a separate variable. Then, save all variables into a variable called `my_sentence`. When you `print(my_sentence)`, the output should look like: `['Actions', 'speak'...]`

**Solve Task 2:**

````python
# YOUR CODE HERE

# saving the string into a variable
my_string = "Actions speak louder than words."

# using slicing to save parts of the string into separate words
word1 = my_string[:7]
word2 = my_string[8:13]
word3 = my_string[14:20]
word4 = my_string[21:25]
word5 = my_string[26:31]

# creating a list that contains all the variables (each variable is a word)
my_sentence = [word1, word2, word3, word4, word5]

# print out the list
print(my_sentence)
````

**Solve task 7:**

````python
# YOUR CODE HERE

# exampe solution (there are many!)

coins = 10 # assuming i have a 10dkk coin
bills = 250 # assuming i have bills of 250dkk in total
ticket_price = 150 # assuming the ticket costs 150 dkk

if coins + bills >= ticket_price: 
# the line above checks whether the sum of coins and bills 
# is AT LEAST equal, or larger than, the ticket price
    print("pay in cash!")
else:
    print("pay by card")
````
