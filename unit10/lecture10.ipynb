<mark style="background-color: lightblue">**x is defined in global scope**</mark>
```python
x = 3 # creating the variable x
```
<mark style="background-color: orange">**input_number and output_number are defined in local scope**</mark>

```python
def add_one(input_number):  # defines the function add_one
```


## Classes, step 5: Adding instance methods

This includes realizing that methods are EITHER accessible ONLY by the class, OR ONLY by the instances. Let's realize it together:

`````python
# do these methods work for the pizzeria, too?
pizzeria = Restaurant()
pizzeria.describe_purpose()
pizzeria.open_restaurant()
pizzeria.close_restaurant()
`````

**Takeaway:**

Global attributes are accessible both by the class and its instances, 

# BUT


static methods work only for the class, NOT for its instances!!

## HOW TO 

define methods that work for instances? >> Enter a new keyword, 

# `self` 

(meaning roughly: "for this instance here!")

`````python
class Restaurant():
    '''This is the Restaurant class.'''

    # global attributes
    purpose = "serving food"
    currency = "dkk"
    lastorder = 22

    # static method (for class Restaurant)
    def describe_purpose():
        print("The purpose of a restaurant is to serve food.")

    # instance method (for instances of the class)
    def open_restaurant(self): # ADDED "SELF" AS FIRST ARGUMENT
        print("The restaurant is open.")

    def close_restaurant(self): # ADDED "SELF" AS FIRST ARGUMENT
        print("The restaurant is closed.")
`````

`````python
# Now, the methods without "self" will work only for the class,
# the methods with "self" will work only for the instances:
Restaurant.describe_purpose() # works only for class
# Restaurant.open_restaurant()
# Restaurant.close_restaurant()
`````

`````python
pizzeria = Restaurant()
pizzeria.describe_purpose()
pizzeria.open_restaurant()
pizzeria.close_restaurant()
`````

`````python
pizzeria = Restaurant()
pizzeria.open_restaurant()
pizzeria.close_restaurant()
`````

## Classes, step 6: adding instance attributes

Levelling up: **How to set attributes flexibly, at instantiation?**

Let's say the only thing that all objects of class Restaurant have in common is the purpose ("serving food"), but we want the instances to have different currencies, and different times for their last order.

Then we can add a **constructor** to our class. The constructor is a function that by convention has the function name `__init__` (two underscores + `init` + two underscores) and **takes as input `self`** (meaning "we're talking about the specific instance!") **and the instance attributes**.

`````python
class Restaurant():
    '''This is the Restaurant class.'''
    purpose = "serving food" # global attribute

    # STATIC METHOD
    def describe_purpose():
        print("The purpose of a restaurant is to serve food.")

    # CONSTRUCTOR - WILL BE RUN AT INSTANTIATION
    def __init__(self, curr, lo):
        self.currency = curr # local attribute, as input argument at instantiation
        self.lastorder = lo # local attribute, as input argument at instantiation
        self.income = 0 # local attribute, set to 0 at instantiation

    # INSTANCE METHOD NR 1
    def open_restaurant(self): # ADDED "SELF" AS FIRST ARGUMENT!!
        print("The restaurant is open.")

    # INSTANCE METHOD NR 2
    def close_restaurant(self): # ADDED "SELF" AS FIRST ARGUMENT!!
        print("The restaurant is closed.")
`````

`````python
Restaurant.purpose # purpose is our only global attribute
# Restaurant.currency # currency is not a global attribute anymore
# Restaurant.lastorder # lastorder is not a global attribute anymore
`````

`````python
# instead, currency and lastorder will be attributes of instances:
pizzeria = Restaurant() # ...and now we need to provide these!
`````

`````python
# instead, currency and lastorder will be attributes of instances:
pizzeria = Restaurant(curr = "euro", lo = 23) # ...and now we need to provide these!
# (we are instantiating an object of class Restaurant, so the __init__ function is run under the hood,
# and it has required input parameters which we need to provide)
`````

`````python
print(pizzeria.currency)
print(pizzeria.lastorder)
`````

`````python
print(Restaurant.purpose) # global attribute: both for class... 
print(pizzeria.purpose) # ... and its instance
print(pizzeria.currency) # local attribute: only for instance, set AT INSTANTIATION
print(pizzeria.lastorder) # local attribute: only for instance, set AT INSTANTIATION
print(pizzeria.income) # local attribute: only for instance, set BY CONSTRUCTOR
`````

`````python
# Restaurant: only global methods (without the "self" keyword)
Restaurant.describe_purpose()
Restaurant.open_restaurant()
Restaurant.close_restaurant()
`````

`````python
# pizzeria (INSTANCE OF Restaurant): only instance methods (WITH the "self" keyword)
pizzeria.describe_purpose()
pizzeria.open_restaurant()
pizzeria.close_restaurant()
`````

# Getters & setters

Again, very short and simplified intro, this is "best practice" (mostly) on how to get & set attributes.

**Getters**: class methods to get the value of an attribute

**Setters**: class methods to set the value of an attribute

## Classes, step 7: Adding "getters" for instance attributes

`````python
class Restaurant(): 

    # getter for class
    def getPurpose():
        return "serving food" 

    def __init__(self, curr, lo):
        self.currency = curr # local attribute
        self.lastorder = lo # local attribute

    # getter for instance (fancy way to access an instance attribute)
    def getCurrency(self):
        return self.currency
`````

`````python
pizzeria = Restaurant("dkk", 22)
print(Restaurant.getPurpose()) # static method
print(pizzeria.getCurrency()) # instance method
`````

`````python
# now we have two ways of accessing the instance attribute "currency":
print(pizzeria.currency) # calling the instance attribute
print(pizzeria.getCurrency()) # calling the getter method
`````
