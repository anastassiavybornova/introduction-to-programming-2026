````python
class Shop():

    '''This is the shop class.'''

    purpose = "selling stuff, making coins"
    alcohol_license = False

    def greet_customer():
        print("Welcome! How can I help you?")

    def __init__(self, shop_name, opens, closes):
        self.shop_name = shop_name
        self.opens = opens
        self.closes = closes
        self.customer_count = 0

    def open_shop(self):
        print("Shop is now open!")
    def close_shop(self):
        print("Shop is now closed!")

    def getName(self):
        return self.shop_name
    def getOpeningTime(self):
        return self.opens
    def getClosingTime(self):
        return self.closes
    def getCustomerCount(self):
        return self.customer_count

    def setName(self, new_name):
        self.shop_name = new_name
    def setOpeningTime(self, new_open_time):
        self.opens = new_open_time
    def setClosingTime(self, new_close_time):
        self.closes = new_close_time
    def updateCustomerCount(self):
        self.customer_count += 1

    def newCustomers(self, N):
        print(f"Welcome to all of my {N} new customers!")
        self.customer_count += N
````
