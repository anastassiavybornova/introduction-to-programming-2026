(1) The size N of input data is the length of the `long_list_of_tasks_to_do`, which *is* used in the code to be looped over, so we already know that runtime is not constant

(2) Steps in the code:

```python
def working_at_the_pool(long_list_of_tasks_to_do):
    print("Here is the pool") # one step
    print("Here am I, working") # one step
    for task in long_list_of_tasks_to_do: # loop runs N times where N is the length of long_list 
        print(f"Doing {task}") # inside the loop there is one step, so the loop gives a total of N*1 = N steps
    print("That's it!") # one last step
```

(3) The total number of steps is, as a function of N (where N=len(long_list_of_tasks_to_do)): N + 3. "Dropping the lower orders" >> we are left with **O(N) (linear time)**

````python
# TIMING THE CODE

# define the function to test
def working_at_the_pool(long_list_of_tasks_to_do):
    print("Here is the pool") # one step
    print("Here am I, working") # one step
    for task in long_list_of_tasks_to_do: # loop runs N times where N is the length of long_list 
        print(f"Doing {task}") # inside the loop there is one step, so the loop gives a total of N*1 = N steps
    print("That's it!") # one last step

Ns = [range(10**2), range(10**3), range(10**4), range(10**5)]
total_times = [] # will save to list to print out separately later

for N in Ns:
    t1 = time.time() # time before
    working_at_the_pool(N)
    t2 = time.time() # time after
    total_time = t2-t1
    total_times.append(total_time)
````

````python
for z in zip(Ns, total_times):
    print(z)
# zip() unpacks both objects at once (so we iterate over Ns and over total_times at the same time)
# see e.g. https://www.geeksforgeeks.org/python/zip-in-python/
````

(1) The input size N is the length of the integer range.

(2) Counting the steps:

```python
def just_making_sure_that_math_still_works(some_integer):
    F = math.factorial(some_integer) # one step
    for i in range(F): # loop that will run N! (N factorial) number of times
        assert 1 + 1 + 1 == 3 # one step inside the loop
        assert 3 * 3 == 9 # one step inside the loop
```

So there is a total of 2 * N! + 1 steps.

(3) Big O, "dropping the coefficients": we are left with **O(N!) (factorial time)**

````python
# timing the code (the runtime increases *very quickly* for larger Ns!)

def just_making_sure_that_math_still_works(some_integer):
    F = math.factorial(some_integer) # one step
    for _ in range(F): # loop that will run N! (N factorial) number of times
        assert 1 + 1 + 1 == 3 # one step inside the loop
        assert 3 * 3 == 9 # one step inside the loop

for N in range(13):
    t1 = time.time()
    just_making_sure_that_math_still_works(N)
    t2 = time.time()
    total_time = t2 - t1
    print(N, total_time)
````

(1) The size of the input data, N, is here the length of the list `chair_list`.

(2) Counting the steps in the code:

```python
def is_close(chair, other_chair):
    """Computes the distance between two points (x1, y1) and (x2, y2) using the
    formula (x1-x2)^2 + (y1-y2)^2 and returning the square root of this."""

    for i in range(len(chair_list)): # iterates N times
        for j in range(i+1, len(chair_list)): # iterates N times
```

Total number of steps: N * N * 3 + 1 = 3 * N ** 2 + 1

(3) Dropping the lower orders and the coefficients, we are left with **O(N ** 2) (polynomial time)**

````python
# TIMING THE CODE

def is_close(chair, other_chair):
    """Computes the distance between two points (x1, y1) and (x2, y2) using the
    formula (x1-x2)^2 + (y1-y2)^2 and returning the square root of this."""
    distance = (chair[0] - other_chair[0])**2 + (chair[1] - other_chair[1])**2 # one step
    distance = math.sqrt(distance) # one step
    return distance <= 1.5 # one step

def has_close_seats(chair_list):
    """Given a list of chairs with their positions [x, y], returns True if there
    is a pair of close chairs, and False otherwise."""
    for i in range(len(chair_list)): # iterates N times
        for j in range(i+1, len(chair_list)): # iterates APPROXIMATELY N times (see findDuplicateBooks example fom AS2, Chapter2)
            if is_close(chair_list[i], chair_list[j]): # 3 steps
                return True # either one more step here...
    return False # ... or one more step here

# generating random chair lists (for worst case runtime: NONE of them are closer than 1.5)

Ns = [100, 500, 1000, 5000, 10**4]
total_times = []

for N in Ns:
    xs = random.choices(range(N*2), k=N)
    ys = random.choices(range(N*3, N*4), k=N)
    chair_list = []
    for z in zip(xs, ys):
        chair_list.append(z)

    t1 = time.time()
    has_close_seats(chair_list)
    t2 = time.time()
    total_time = t2 - t1
    total_times.append(total_time)
    print(N, total_time)
````
