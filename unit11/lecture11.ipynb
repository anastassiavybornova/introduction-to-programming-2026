```python
# timeit module, .timeit() function
timeit.timeit('print("hello world")', number=1) # by default, number=1000000 (1 million runs)
```

```python
# timeit module, .timeit() function
timeit.timeit('1+1') # seconds to do this *number* of times (by default, number=1.000.000)
```

```python
timeit.timeit('1+1', number=1) # seconds to do this *number* of times (by default, number=1.000.000)
```

```python
timeit.timeit('[1]*10') # creating a list of length 10... a million times
```

```python
# how long does it take to create a list of length N?
# using the time.time() function:
N = 10
t1 = time.time()
list_of_zeros = [0] * N
t2 = time.time()
total_time = t2-t1
print(f"list of length {N}, total time {total_time}")
```

```python
# how long does it take to create a list of length N? // for different values of N:
for N in [10, 100, 1000, 10000, 100000]:
    t1 = time.time()
    list_of_zeros = [0] * N
    t2 = time.time()
    print(N, t2-t1)
```

## Counting steps in code

```python
# counting the steps in the code
def iLoveBooks(books): # n = len(books)
    for i in range(10):
        print("I love books!")
        print("Books are great!")

```

```python
# counting steps in the code
def readingList(books): # n = len(books)
    print("Books I will read:")
    numberOfBooks = 0
    for book in books:
        print(book)
        numberOfBooks += 1
    print(numberOfBooks, "books total.")

```

```python
# counting steps in the code
def quadraticExample(someData): # n = len(someData)
    for i in someData:
        for j in someData:
            print("Something")
            print("Something")
            print("Something")
```

```python
# counting steps in the code
def linearExample(someData): # n = len(someData)
    for i in someData:
        for k in range(100):
            print("Something")
```

```python
# counting steps in the code
def countBookPoints(books):
    points = 0
    for book in books:
        points += 1
    for book in books:
        if "by Al Sweigart" in book:
            points += 1
    return points
```

```python
# counting steps in the code
def books_in_right_place(books):
    for i in books:
        for j in sorted(books):
            if i == j:
                print("Book is in right place:", books[i])
                print("Its position in the list is:", i)
```
